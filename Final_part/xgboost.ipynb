```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier
```

```python
important_features = pd.read_csv("../Final_data/feature_importances.csv")
df = pd.read_csv("../Final_data/songs_features.csv")
```

```python
num_features = [2, 4, 6, 8, 12, 15, 20]

for n in num_features:
    top_n_features = important_features.head(n)["feature"].tolist()
    filt_df = df[top_n_features + ["label"]].copy()
    X = filt_df.drop("label", axis=1)
    y = filt_df["label"]
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)
    model = XGBClassifier()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Accuracy for top {n} features: {accuracy}")
```
