```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../Data/eeg_features.csv")
```

```python
transactions= pd.DataFrame()

for zone in ["F4", "C4", "E2"]:
    for wave in ["alpha", "delta", "theta", "gamma", "beta"]:
        col_name = f"{zone}_bp_{wave}"
        transactions[col_name] = df[col_name] > df[col_name].median()

```

```python
col_name = "F4_bp_alpha"
 
np.mean(df[col_name] > df[col_name].median())
```

```python
transactions
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
frequent_itemsets = apriori(transactions, 
                            min_support=0.2, 
                            use_colnames=True)

rules = association_rules(frequent_itemsets, 
                          metric="confidence", 
                          min_threshold=0.2)
```

```python
rules.sort_values(by = "support", ascending=False).head(15)
```
